### Dimensionality Reduction

```python
# import necessary libs.

import numpy as np
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
```

### Loading dataset ...

```python
X, y = load_iris(return_X_y=True)
X.shape, y.shape
```

### Visualizing the data

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(121)
plt.scatter(X[:, 0], X[:, 1], c=y, marker='+')
plt.grid(True)
plt.subplot(122)
plt.scatter(X[:, 2], X[:, 3], c=y, marker='o')
plt.grid(True)
plt.show()
```

### Applying PCA and transforming features

```python
pca_2d = PCA(n_components=2)
X_transformed = pca_2d.fit_transform(X)
X_transformed.shape
```

### Visualizing again

```python
plt.scatter(X_transformed[:, 0], X_transformed[:, 1], c=y, marker='+')
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.grid(True)
plt.show()
```
